# Pre-procesamiento de datos de notas clínicas

<div style="text-align: justify;">
El preprocesamiento de datos es una etapa fundamental en el análisis de textos (procesamiento del lenguaje natural, o NLP) que implica preparar y limpiar los datos para que puedan ser procesados de manera más eficiente por modelos o algoritmos. El objetivo principal es transformar el texto en un formato que sea más fácil de analizar, eliminar información irrelevante o redundante y destacar las características importantes que serán útiles en el análisis posterior.

En el proceso de preprocesamiento se realizaron los siguientes pasos:

Eliminación de caracteres y signos: Se eliminaron caracteres especiales, como símbolos de puntuación y otros signos (comas, puntos, paréntesis, entre otros) que no aportan valor semántico para el análisis. Esto ayuda a simplificar el texto, ya que estos signos no suelen contribuir al significado central de las palabras.

Tokenización: La tokenización consiste en dividir el texto en sus componentes más pequeños, conocidos como tokens. En este caso, se tokenizó el texto para obtener las palabras individuales, lo que permite tratarlas de manera independiente para su posterior análisis.

Eliminación de stopwords: Las stopwords son palabras comunes en un idioma (como "y", "el", "de", "un") que suelen aparecer con mucha frecuencia pero no aportan un significado relevante al contenido del texto. Eliminarlas permite que el análisis se centre en las palabras que tienen mayor valor informativo.
</div>

# Notas clínicas, preprocesamiento y obtención de datos

```python
def preprocesamiento(contenido):
        # Eliminar valores alfanuméricos
        contenido = re.sub(r'[^a-zA-ZáéíóúÁÉÍÓÚñÑ\s]', '',contenido)

        # Convertir a minusculas
        contenido = contenido.lower()

        # Tokenizar
        tokens = word_tokenize(contenido)

        # Eliminar las stop words
        stop_words = set(stopwords.words('spanish'))

        tokens = [palabra for palabra in tokens if palabra not in stop_words]

        # lemantizar
        # lematizados = []
        # for palabra in tokens:
        #     doc = nlp(palabra)
        #     lema = doc[0].lemma_ if doc else palabra  # Obtener el lema de la primera forma encontrada
        #     lematizados.append(lema)
        # return(lematizados)
        return (tokens)
```

```python
def contar_palabras(contenido):
    palabras = contenido.split()
    num_palabras = len(palabras)
    return num_palabras
```

```python
def frecuencia_palabras(texto_preprocesado):
        # Contar la frecuencia de las palabras
        contador = Counter(texto_preprocesado)
        # Obtener las 20 palabras más repetidas
        palabras_repetidas = contador.most_common(20)
        return(palabras_repetidas)
```

```python
def busqueda_de_palabras(texto_preprocesado, archivo):
    # Definir las palabras clave a buscar
    palabras_clave = ["ansiedad", "depresión", "estrés"]
    resultados = {}
    total_coincidencias = 0

    for palabra in palabras_clave:
        # Contador para almacenar el número de coincidencias
        contador_coincidencias = 0
        # Iterar sobre cada palabra en el texto preprocesado
        for palabra_nota in texto_preprocesado:
            # Buscar coincidencias en cada palabra de la nota clínica
            if re.search(palabra, palabra_nota):
                contador_coincidencias += 1
                total_coincidencias += 1
        # Almacenar el número de coincidencias para esta palabra clave
        resultados[palabra] = contador_coincidencias
    resultados["total"] = total_coincidencias
    resultados["nota"] = archivo
    return resultados

def sumar_coincidencias(lista_diccionarios):
    # Inicializar contadores para cada palabra clave
    total_ansiedad = 0
    total_depresion = 0
    total_estres = 0
    total_general = 0

    # Iterar sobre cada diccionario en la lista
    for diccionario in lista_diccionarios:
        # Sumar las coincidencias para cada palabra clave
        total_ansiedad += diccionario['ansiedad']
        total_depresion += diccionario['depresión']
        total_estres += diccionario['estrés']
        total_general += diccionario['total']

    # Retornar los totales acumulados como un diccionario
    return {
        'ansiedad': total_ansiedad,
        'depresión': total_depresion,
        'estrés': total_estres,
        'total': total_general
    }

def notas_con_palabras(lista_diccionarios):
    # Inicializar contadores para cada palabra clave
    total_ansiedad = 0
    total_depresion = 0
    total_estres = 0
    total_general = 0

    # Iterar sobre cada diccionario en la lista
    for diccionario in lista_diccionarios:
        # Sumar las coincidencias para cada palabra clave
        if diccionario['total']>0:
            total_general += 1
        if diccionario['ansiedad']>0:
            total_ansiedad += 1
        if diccionario['depresión']>0:
            total_depresion += 1
        if diccionario['estrés']>0:
            total_estres += 1

    # Retornar los totales acumulados como un diccionario
    return {
        'ansiedad': total_ansiedad,
        'depresión': total_depresion,
        'estrés': total_estres,
        'total': total_general
    }
```

```python

# Ruta de la carpeta que contiene los archivos que quieres analizar
carpeta_origen = '/content/drive/MyDrive/Pasantía_mexico/notas_clinicas_corpus_distemist'

# Ruta de la carpeta donde se guardarán los archivos preprocesados
carpeta_destino = '/content/drive/MyDrive/notas_preprocesadas'

# Crear la carpeta de destino si no existe
if not os.path.exists(carpeta_destino):
    os.makedirs(carpeta_destino)

# Inicialización de listas para el almacenamiento de datos
palabras = []
palabras_frecuencia = []
conteo_palabras = []
enfermedades_en_notas = []
trastornos_en_notas = []

# Inicializar contadores de trastornos
conteo_trastornos = {"ansiedad": 0, "estrés": 0, "depresión": 0}

i = 1

# Iterar a través de archivos en carpeta_origen y subcarpetas
for ruta_directorio, _, archivos in os.walk(carpeta_origen):
    for archivo in archivos:

        # Obtener la ruta completa del archivo
        ruta_archivo = os.path.join(ruta_directorio, archivo)

        # Comprobar si el elemento es un archivo (no una carpeta)
        if os.path.isfile(ruta_archivo):
            # Leer el contenido del archivo
            with open(ruta_archivo, 'r', encoding='utf-8') as file:
                contenido = file.read()
                #Pre-procesar el texto
                texto_preprocesado = preprocesamiento(contenido)

                #Obtener la frecuencia de palabras tokenizadas
                palabras_repetidas = frecuencia_palabras(texto_preprocesado)
                palabras_frecuencia.append(palabras_repetidas)

                # Contar palabras en la nota clínica
                num_palabras = contar_palabras(contenido)
                conteo_palabras.append(num_palabras)

                # Buscar enfermedades en el contenido preprocesado
                enfermedades_encontradas = [enfermedad for enfermedad in enfermedades if enfermedad in texto_preprocesado]
                if len(enfermedades_encontradas) > 1:
                    print(f'Nota con más de una enfermedad: {archivo}, Enfermedades: {enfermedades_encontradas}')
                enfermedades_en_notas.append(enfermedades_encontradas)

                ''' # Buscar trastornos en el contenido preprocesado
                conteo_individual_trastornos = {trastorno: texto_preprocesado.count(trastorno) for trastorno in trastornos}
                trastornos_en_notas.append(conteo_individual_trastornos) '''

                palabras_encontradas = busqueda_de_palabras(texto_preprocesado, archivo)
                palabras.append(palabras_encontradas)

                ''' # Sumar al conteo total de trastornos
                for trastorno, conteo in conteo_individual_trastornos.items():
                    conteo_trastornos[trastorno] += conteo
 '''
                # Convertir la lista de tokens en una cadena antes de escribir
                texto_preprocesado_str = ' '.join(texto_preprocesado)

                # Guardar el texto preprocesado en la carpeta de destino
                ruta_archivo_destino = os.path.join(carpeta_destino, archivo)
                with open(ruta_archivo_destino, 'w', encoding='utf-8') as archivo_salida:
                    archivo_salida.write(texto_preprocesado_str)

            print(f"Archivo {i}, nota {archivo}")
            i += 1

print("Preprocesamiento completado. Los archivos preprocesados se han guardado en:", carpeta_destino)
```

# Frecuencia de palabras repetidas

```python
# Inicializa un diccionario para llevar un registro de las repeticiones totales por palabra
repeticiones_totales_por_palabra = {}

# Itera a través de las listas de tuplas y calcula la frecuencia total de cada palabra
for lista_de_tuplas in palabras_frecuencia:
    # Itera a través de las tuplas en la lista actual
    for tupla in lista_de_tuplas:
        palabra, frecuencia = tupla
        if palabra in repeticiones_totales_por_palabra:
            repeticiones_totales_por_palabra[palabra] += frecuencia
        else:
            repeticiones_totales_por_palabra[palabra] = frecuencia

# Lematiza las palabras en el diccionario
lemmatized_dict = {}

for word, count in repeticiones_totales_por_palabra.items():
    # Aplica lematización a la palabra
    doc = nlp(word)
    # Obtiene la forma base (lema) de la palabra
    lemma = doc[0].lemma_
    # Agrega el lema y el número de repeticiones al diccionario lematizado
    if lemma in lemmatized_dict:
        lemmatized_dict[lemma] += count
    else:
        lemmatized_dict[lemma] = count

# Ordena el diccionario por valor (frecuencia) de mayor a menor
diccionario_ordenado = dict(sorted(lemmatized_dict.items(), key=lambda item: item[1], reverse=True))

# Imprime las 10 primeras palabras y sus frecuencias
contador = 0
for palabra, repeticiones in diccionario_ordenado.items():
    print("Palabra:")
    print(f"{palabra}")
    print("Repetición:")
    print(f"{repeticiones}")
    contador += 1
    if contador >= 10:
        break
```
